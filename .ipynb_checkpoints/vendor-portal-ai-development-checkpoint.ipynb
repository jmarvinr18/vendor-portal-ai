### create_file tool

```python
s3 = boto3.client("s3", region_name=os.getenv("AWS_REGION", "ap-southeast-1"))

def _latin1(text: str) -> str:
    repl = {"\u2018": "'", "\u2019": "'", "\u201c": '"', "\u201d": '"',
            "\u2013": "-", "\u2014": "-", "\u2022": "-", "\u2026": "..."}
    for k, v in repl.items():
        text = text.replace(k, v)
    return text.encode("latin-1", "replace").decode("latin-1")

@tool(parse_docstring=True)
def create_pdf_report(title: str, content: str) -> str:
    """Create a PDF report and return a download link. Use this whenever the user asks to export, save, or download something as a PDF.

    Args:
        title: The report title shown at the top of the first page.
        content: The full report body as plain text. Separate paragraphs with newlines.
    """
    bucket = os.getenv("REPORTS_BUCKET")
    if not bucket:
        return "PDF could not be created: REPORTS_BUCKET is not configured."

    path = f"/tmp/{uuid.uuid4().hex}.pdf"
    pdf = FPDF(orientation="P", unit="mm", format="A4")
    pdf.set_auto_page_break(auto=True, margin=15)
    pdf.add_page()
    pdf.set_font("Helvetica", style="B", size=20)
    pdf.multi_cell(0, 15, _latin1(title), align="C")
    pdf.ln(10)
    pdf.set_font("Helvetica", size=12)
    for para in content.split("\n"):
        if para.strip():
            pdf.multi_cell(0, 7, _latin1(para.strip()))
            pdf.ln(4)

    key = f"reports/{os.path.basename(path)}"
    try:
        pdf.output(path)
        s3.upload_file(path, bucket, key, ExtraArgs={"ContentType": "application/pdf"})
        url = s3.generate_presigned_url("get_object",
                                        Params={"Bucket": bucket, "Key": key},
                                        ExpiresIn=3600)
    except Exception as e:
        # Goes to the runtime's CloudWatch log; the model only gets a short message
        print(f"create_pdf_report failed (bucket={bucket}, key={key}): {e!r}")
        return "PDF could not be created because of a storage error. Please try again later."
    finally:
        if os.path.exists(path):
            os.remove(path)
    return f"PDF created. Download link (valid 1 hour): {url}"
```

```python
from langchain_aws.retrievers import AmazonKnowledgeBasesRetriever
from botocore.config import Config

def get_retriever():

    kb_client = boto3.client(
        "bedrock-agent-runtime",
        region_name="ap-southeast-1",
        config=Config(
            retries={"max_attempts": 6, "mode": "adaptive"},  # standard also fine
            tcp_keepalive=True,      # detect dead sockets instead of reusing them
            connect_timeout=5,
            read_timeout=30,
            max_pool_connections=10,
        ),
    )    
    retriever = AmazonKnowledgeBasesRetriever(
        knowledge_base_id="DTDQDQUF9E",
        client=kb_client,
        region_name="ap-southeast-1",
        retrieval_config={
            "vectorSearchConfiguration": {"numberOfResults": 10}
        },        
    )
    return retriever
```

### web_browse tool

```python
# from langchain_community.tools.tavily_search import TavilySearchResults
from langchain_tavily import TavilySearch
from langgraph.prebuilt import ToolNode
# from dotenv import load_dotenv
# from config import load_secrets

# load_dotenv()
# load_secrets()

def browse(query: str):
    """
    Search the web for up-to-date information and return the top results (with images).
    Use this for current events or anything not found in the internal documents.
    """

    print(f"QUERY FOR BROWSING: {query}")

    tool = TavilySearch(max_results=2, include_images=True)


    return tool.invoke({"query": query})

def create_tool_node(tools):
    """
    Creates and returns a tool node for the graph
    """
    return ToolNode(tools=tools)
```

```python
def latest_question(messages):
    """The most recent user message. messages[0] is the first question of the whole thread."""
    for message in reversed(messages):
        if isinstance(message, HumanMessage):
            return message.content
    return ""


def latest_tool_results(messages):
    """The ToolMessages produced by the last tool-calling step."""
    results = []
    for message in reversed(messages):
        if not isinstance(message, ToolMessage):
            break
        results.append(message)
    return list(reversed(results))
```

## NODES

```python
from pydantic import BaseModel, Field
from typing import Annotated, Sequence, Literal
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
from langchain_core.messages import BaseMessage, HumanMessage, AIMessage, ToolMessage
from langchain_core.output_parsers import StrOutputParser
```

```python
def invoke_agent(state):
    """
        Invokes the agent model to generate a response based on the current state. Given the question, it will
        decide to retrieve using the retriever tool, or simply end.
        
        Args:
            state (messages): The current state

        Returns:
            dict: The updated state with the agent response appended to messages
    """
    print("---CALL AGENT--")

    messages = state["messages"]

    # model = model.bind_tools(tools)

    llm_with_tools = llm.bind_tools(tools)

    response = llm_with_tools.invoke(messages)

    return {"messages": [response]}
```

```python
def route_tools(state) -> Literal["generate", "rewrite", "agent"]:
    """
        Only retriever results get graded. Other tools (browse, math, PDF) go back to the
        agent so it can keep working with their output.
    """
    results = latest_tool_results(state["messages"])
    if any(m.name == RETRIEVER_TOOL_NAME for m in results):
        return grade_documents(state)
    return "agent"
```

```python
def grade_documents(state) -> Literal["generate","rewrite"]:
    """
        Determines whether the retrieved documents are relevant to the question.
        Args:
            state (messages): The current state
        Returns:
            str: A decision for whether the documents are relevant or not
    """

    print("--CHECK RELEVANCE--")

    # Data model
    class grade(BaseModel):
        """Binary score for relevance check."""
        binary_score: str = Field(description="Relevance score 'yes' or 'no'")


    ## LLM with tool and validation
    llm_with_tool = llm.with_structured_output(grade)

    # Prompt
    prompt=PromptTemplate(
        template="""You are a grader assessing relevance of a retrieved document to a user question. \n
            Here is the retrieved document: \n\n {context} \n\n
            Here is the user question: {question} \n
            If the document contains keyword(s) or semantic meaning related to the user question, grade it as relevant. \n
            Give a binary score 'yes' or 'no' score to indicate whether the document is releavant to the question.
        """,
        input_variables=["context", "question"],
    )

    # Chain
    chain = prompt | llm_with_tool
    messages = state["messages"]

    question = latest_question(messages)
    docs = "\n\n".join(m.content for m in latest_tool_results(messages)
                       if m.name == RETRIEVER_TOOL_NAME)

    scored_result=chain.invoke({"question": question, "context":docs})

    score = scored_result.binary_score

    if score == "yes":
        print("---DECISION: DOCS RELEVANT---")
        return "generate"
    else:
        print("---DECISION: DOCS NOT RELEVANT---")
        print(score)
        return "rewrite"
```

```python
def rewrite(state):
    """
    Transform the query to produce a better question.
    Args:
        state (messages): The current state
    Returns:
        dict: The updated state with re-phrased question
    """
    print("---TRANSFORM QUERY---")
    messages = state["messages"]
    question = latest_question(messages)
    msg = [
        HumanMessage(
            content=f""" \n
                Look at the input and try to reason about the underlying semantic intent / meaning. \n
                Here is the initial question:
                \n ------- \n
                {question}
                \n ------- \n
                Reply with only the improved question, nothing else.
            """,
        )
    ]
    # Grader
    response = StrOutputParser().invoke(llm.invoke(msg))
    # A HumanMessage so the agent treats it as the question to answer, not as its own reply
    return {"messages": [HumanMessage(content=response)]}
```

```python
    def generate(state):
        """
            Generate answer
            Args:
                state (messages): The current state
            Returns:
                dict: The updated message
        """
        print("---GENERATE---")
        messages = state["messages"]
        question = latest_question(messages)
        docs = "\n\n".join(m.content for m in latest_tool_results(messages))

        # Prompt
        prompt = ChatPromptTemplate.from_messages([
            ("human", """You are an assistant for question-answering tasks. Use the following pieces of retrieved context to answer the question. If you don't know the answer, just say that you don't know. Use three sentences maximum and keep the answer concise.
        Question: {question} 
        Context: {context} 
        Answer:""")
        ])


        # Post-processing
        def format_docs(docs):
            return "\n\n".join(doc.page_content for doc in docs)

        # Chain
        rag_chain = prompt | llm | StrOutputParser()

        # Run
        response = rag_chain.invoke({"context":docs, "question": question})

        return {"messages": [AIMessage(content=response)]}
```

## GRAPHS

```python
from langgraph.prebuilt import tools_condition
from langgraph.checkpoint.memory import MemorySaver
from IPython.display import Image, display
```

```python
graph = StateGraph(AgentState)

tools = [retriever_tool(), add, multiply, browse, create_pdf_report]

# Nodes
# Define the nodes we will cycle between
graph.add_node("agent", invoke_agent)

# agent
# A failing tool returns an error ToolMessage to the agent instead of aborting the stream
toolnode = ToolNode(tools, handle_tool_errors=True)


graph.add_node("retrieve_tool", toolnode)

# retrieval
# Re-writing the question
graph.add_node("rewrite", rewrite)
graph.add_node("generate", generate)

# Generating a response after we know the documents are relevant
# Call agent node to decide to retrieve or not
graph.add_edge(START, "agent")

# Decide whether to retrieve
graph.add_conditional_edges("agent",
                                 # Assess agent decision
                                 tools_condition,
                                 {
                                     # Translate the condition outputs to nodes in our graph
                                     "tools": "retrieve_tool",
                                     END: END,
                                 }
                                 )
# Edges taken after the `action` node is called.
graph.add_conditional_edges(
    "retrieve_tool", route_tools)
graph.add_edge("generate", END)
graph.add_edge("rewrite", "agent")

graph_builder= graph.compile(checkpointer=MemorySaver())
```
